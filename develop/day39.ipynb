```python
def dft(y):
    N = len(y)
    c = np.zeros(N//2+1, complex)
    for k in range(N//2+1):
        for n in range(N):
            c[k] += y[n]*np.exp(-2j*np.pi*k*n/N)
    return(c)
```

```python
x0 = np.linspace(0,10, 500)
y0 = np.sin(2*np.pi*x0) + np.sin(5*np.pi*x0) + np.sin(10*np.pi*x0)
```

```python
c = dft(y0)
```

```python
fig0, ax0 = plt.subplots()
ax0.plot(c.real)
```

```python
cfft = np.fft.rfft(y0)
dx = x0[1]-x0[0]
w = np.fft.rfftfreq(500, d=dx)
```

```python
fig1, ax1 = plt.subplots()
ax1.plot(w, cfft.real)
```

```python
L = 10
x = np.linspace(0, 3*L, 500)
y = (x % L)
```

```python
fig2, ax2 = plt.subplots()
ax2.plot(x, y)
```

```python
cfft = np.fft.rfft(y)
dx = x[1]-x[0]
w = np.fft.rfftfreq(500, d=dx)

fig4, ax4 = plt.subplots()
ax4.plot(w, np.sqrt(cfft*cfft.conjugate())) # here i am plotting the magnitude of the fourier transform
```

```python
y5 = np.fft.irfft(cfft)
```

```python
fig5, ax5 = plt.subplots()
ax5.plot(x, y5, '.')
ax5.plot(x, y)
```
