```python
h = float(input("Enter the height of the tower: "))
t = float(input("Enter the time interval: "))
g = 9.81

y = h - 1/2*g*t**2
if y>0:
    print('The height of the rock is ', y, 'meters')
else:
    print('The height of the rock is 0 meters.')
```
