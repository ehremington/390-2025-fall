These methods all go by the name Runge-Kutta. Newton's method is actually 1st order Runge-Kutta. 2nd order Runga-Kutta and 4th order Runga-Kutta are what we are going to talk about today.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Euler's Method

# define the differential equation
def f(x_, y_): # dy/dx
    return(-y_**3+np.sin(x_))

# define the boundary condition
a = 0.0
b = 10.0
N = 1000
dx = (b-a)/N

y = 0.0 # initial condition

xpoints = np.arange(a, b, dx)
ypoints = []

for i in xpoints:
    ypoints.append(y)
    y = y + f(i, y) * dx

```

```python
fig0, ax0 = plt.subplots()
ax0.plot(xpoints, ypoints)
```

```python
# Runge-Kutta 2nd order

y = 0.0 # initial condition

xpoints = np.arange(a, b, dx)
ypoints = []

for i in xpoints:
    ypoints.append(y)
    k1 = dx*f(i,y)
    k2 = dx*f(i+1/2*dx, y+1/2*k1)
    y = y + k2

```

```python
# Runge-Kutta 4th order

y = 0.0 # initial condition

xpoints = np.arange(a, b, dx)
ypoints = []

for i in xpoints:
    ypoints.append(y)
    k1 = dx*f(i,y)
    k2 = dx*f(i+1/2*dx, y+1/2*k1)
    k3 = dx*f(i+1/2*dx, y+1/2*k2)
    k4 = dx*f(i+dx, y+k3)
    y = y + (k1+2*k2+2*k3+k4)/6
```

```python
fig2, ax2 = plt.subplots()
ax2.plot(xpoints, ypoints, '+')
```
