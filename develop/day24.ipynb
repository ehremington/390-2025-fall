## herein

We have worked on Euler's method and have a decent way to do that. But there are other methods that work even better and don't take that much more calculation to do. These methods all go by the name Runge-Kutta. Newton's method is actually 1st order Runge-Kutta. 2nd order Runga-Kutta and 4th order Runga-Kutta are what we are going to talk about today.

```python
%matplotlib ipympl
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Euler's Method

# define the differential equation
def f(x_, t_): # dx/dt
    # return(-x_**3+np.sin(t_))
    return(t_**2+t_-1)

# define the boundary condition
a = 0.0
b = 10.0
N = 50
dt = (b-a)/N

x = 1.0 # initial condition

tpoints = np.arange(a, b, dt)
xpointsE = []

for i in tpoints:
    xpointsE.append(x)
    x = x + f(x, i) * dt

fig0, ax0 = plt.subplots()
ax0.plot(tpoints, xpointsE)
```

```python
# Runge-Kutta 2nd order

x = 1.0 # initial condition

tpoints = np.arange(a, b, dt)
xpointsRK2 = []

for i in tpoints:
    xpointsRK2.append(x)
    k1 = dt*f(x, i)
    k2 = dt*f( x+1/2*k1, i+1/2*dt)
    x = x + k2

```

```python
# Runge-Kutta 4th order

x = 1.0 # initial condition

tpoints = np.arange(a, b, dt)
xpointsRK4 = []

for i in tpoints:
    xpointsRK4.append(x)
    k1 = dt*f(x, i)
    k2 = dt*f(x+1/2*k1, i+1/2*dt)
    k3 = dt*f(x+1/2*k2, i+1/2*dt)
    k4 = dt*f(x+k3, i+dt)
    x = x + (k1+2*k2+2*k3+k4)/6

fig2, ax2 = plt.subplots()
ax2.plot(tpoints, xpointsRK4, '+--')
```
