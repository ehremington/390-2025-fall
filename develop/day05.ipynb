
1. make a factorial function
2. make it return the factorial for any number (use a for loop!)
3. don't test it too high!

````python
def factorial(x):
    f = 1
    if x<0:
        print('undefined')
    else:
        for i in range(1,x+1):
            f = f*i

        return(f)
````

````python
y = factorial(5)
````

````python
def distance(x,y,z):return((x**2+y**2+z**2)**(1/2))
````

````python
def sph(r, theta, phi):
    from math import cos, sin
    x = r*sin(theta)*cos(phi)
    y = r*sin(theta)*sin(phi)
    z = r*cos(theta)
    return(x,y,z)
    
````

````python
from math import pi

sph(1,pi/2, 0)
````
