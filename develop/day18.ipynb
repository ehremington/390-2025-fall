```python
x = 2
error = 1
count = 0 
while abs(error)>0.00001:
    x1 = np.sqrt(1-np.log(x))
    error = x1 - x
    x = x1
    count = count + 1 
    if count > 100000:
        print('i couldnt find an answer so ignore this')
        break
print(x)
```
