```python
g = 9.8 # N/kg
m = 1 # kg
c = 0.1 # crazy air resistance units


# define the differential equation
def f(r, t): # don't forget r is a vector!
    x, y, z, vx, vy, vz = r
    xdot = vx
    ydot = vy
    zdot = vz
    vxdot = 0
    vydot = -g
    vzdot = 0
    return(np.array([xdot, ydot, zdot, vxdot, vydot, vzdot]))

# define the boundary condition
a = 0.0
b = 20.0
N = 1000
dt = (b-a)/N

# Runge-Kutta 4th order

# r = np.array([x_i, y_i, z_i, vx_i, vy_i, vz_i])
r = np.array([0, 0, 0, 100, 100, 0]) # initial condition

tpoints = np.arange(a, b, dt)

xpoints = []
ypoints = []
zpoints = []
vxpoints = []
vypoints = []
vzpoints = []

for i in tpoints:
    xpoints.append(r[0])
    ypoints.append(r[1])
    zpoints.append(r[2])
    vxpoints.append(r[3])
    vypoints.append(r[4])
    vzpoints.append(r[5])
    k1 = dt*f(r, i)
    k2 = dt*f(r+1/2*k1, i+1/2*dt)
    k3 = dt*f(r+1/2*k2, i+1/2*dt)
    k4 = dt*f(r+k3, i+dt)
    r = r + (k1+2*k2+2*k3+k4)/6

fig0, (ax0, ax1) = plt.subplots(1, 2, figsize=(7,3))
ax0.plot(tpoints, xpoints)
ax1.plot(tpoints, ypoints)
# ax0.plot(tpoints, vpoints)
ax0.grid(True)
ax1.grid(True)
fig0.tight_layout()
```
