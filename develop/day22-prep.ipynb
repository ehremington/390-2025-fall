```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Euler's Method

def f(x,t):
    # return(-x**3 + np.sin(t))
    return(2*x/t)
# return(2*t)

# define boundary conditions

a = 1 # starting point
b = 3.0 # ending point
N = 1000 # number of points between a and b
dt = (b-a)/N

x = 1.0 # initial condition


tpointsE = np.arange(a, b, dt)
xpointsE = []

for t in tpointsE:
    xpointsE.append(x)
    x = x + dt*f(x,t)
```

```python
fig, ax = plt.subplots()
ax.plot(tpointsE, xpointsE)
#ax.plot(tpointsE, tpointsE**2)
```

```python
# Runge-Kutta 4nd order

#def f(x,t):
#    return(-x**3 + np.sin(t))

# define boundary conditions

#a = 0.0 # starting point
#b = 3.0 # ending point
#N = 100 # number of points between a and b
#dt = (b-a)/N

x = 1.0 # initial condition


tpointsRK4 = np.arange(a, b, dt)
xpointsRK4 = []

for t in tpointsRK4:
    xpointsRK4.append(x)
    k1 = dt*f(x,t)
    k2 = dt*f(x+0.5*k1,t+0.5*dt)
    k3 = dt*f(x+0.5*k2,t+0.5*dt)
    k4 = dt*f(x+k3, t+dt)
    x = x + (k1+2*k2+2*k3+k4)/6
```

```python
fig1, ax1 = plt.subplots()
ax1.plot(tpointsRK4, xpointsRK4)
```
