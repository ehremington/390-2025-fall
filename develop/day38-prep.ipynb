```python
x = np.linspace(0,10,11)
y = x

L = 10
x = np.linspace(0, 3*L, 500)   # show three periods bc it looks cooler
y = (x % L)                    # periodic sawtooth

fig0,ax0=plt.subplots()
ax0.plot(x, y)
```

so how can i make a sawtooth of that quickly. 

```python
def saw(x, L=2*np.pi):
    return (x % L)/L   # normalized to [0,1]

fig1, ax1 = plt.subplots()
ax1.plot(x,saw(x))
```
