````python
import numpy as np
import matplotlib.pyplot as plt
from gaussxw import gaussxwab
````

## first let's plot some things

Let's plot the original integral just to get an idea of what we are looking at.

````python
fig0, ax0 = plt.subplots()

x = np.linspace(0,10,100)
y = x**3/(np.exp(x)-1)

ax0.plot(x, y)
````

So that looks like it would have a value of maybe 10 or less. Let's look at the modified function next that we would integrate from 0 to 1.

````python
fig1, ax1 = plt.subplots()

z= np.linspace(0.001,.99,100)
f = z**3/(1-z)**5/(np.exp(z/(1-z))-1)

ax1.plot(z, f)
````

Ok so now we are ready to do some integration since we have an expectation of what we should get. Both of these look to be integrals that will work out to be about 10 or maybe less.

````python
# constants

kB = 1.38*10**(-23)
c = 3*10**8
hbar = 6.626e-34/2/np.pi
````

````python
def f(z):
    return(z**3/(1-z)**5/(np.exp(z/(1-z))-1))

N = 50
x, w = gaussxwab(N, 0.001, .99)
s = 0.0
for k in range(N):
    s = s + w[k]*f(x[k])
print(s)
````

Ok so that value seems very plausible given the functions that we looked at up above. Now what we need to do is multiply and divide correctly to get there.

$$\sigma = \frac{k_B^4}{4 \pi^2 c^2 \hbar^3} \times 6.493939...$$

This should be close to $ \sigma = 5.6704 \times 10^{-8} \; W m^{−2}⋅K^{−4}$ 

````python
print(s*(kB**4)/4/(np.pi**2)/(c**2)/(hbar**3))
````
