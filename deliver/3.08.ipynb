```python
df = pd.read_csv('../data/millikan.txt', sep=' ', names=['x', 'y'])
```

```python
df
```

```python
n = len(df)

ex = 1/n*np.sum(df['x'])
ey = 1/n*np.sum(df['y'])
exx = 1/n*np.sum(df['x']**2)
exy  = 1/n*np.sum(df['x']*df['y'])

m = (exy - ex*ey)/(exx - ex**2)
c = (exx*ey - ex*exy)/(exx - ex**2)

print('slope = ', m)
print('y intercept = ', c)

df['fit'] = m*df['x'] + c

fig0, ax0 = plt.subplots()

ax0.plot(df['x'], df['y'], 'o')
ax0.plot(df['x'], df['fit'])
```
