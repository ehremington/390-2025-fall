```python
import numpy as np
import matplotlib.pyplot as plt
```

## part a

```python
theta = np.linspace(0,2*np.pi)

x = 2*np.cos(theta)+np.cos(2*theta)
y = 2*np.sin(theta) - np.sin(2*theta)
```

```python
y
```

```python
fig0 = plt.figure()
ax0 = fig0.add_subplot()

ax0.plot(x, y)
```

## part b

```python
theta = np.linspace(0,10*np.pi, 1000)

r = theta**2

x = r*np.cos(theta)
y = r*np.sin(theta)
```

```python
fig1 = plt.figure()
ax1 = fig1.add_subplot(aspect="equal")

    
ax1.plot(x,y)
```

## part c

```python
theta = np.linspace(0,24*np.pi, 1000)

r = np.exp(np.cos(theta))-2*np.cos(4*theta)+np.sin(theta/12)

x = r*np.cos(theta)
y = r*np.sin(theta)
```

```python
fig2, ax2 = plt.subplots(subplot_kw = {'aspect':'equal'})

ax2.plot(x, y)
plt.show()
```
