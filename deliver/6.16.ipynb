```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
G = 6.674*10**-11
M = 5.974*10**24
m = 7.348*10**22
R = 3.844*10**8
w = 2.662*10**-6
```

```python
def f(x):
    return(G*M/x**2-G*m/(R-x)**2-w**2*x)
```

after a long time of zooming in i finally found it...

```python
fig, ax = plt.subplots()
x_ = np.linspace(3.2e8,3.3e8,1000)
ax.plot(x_, f(x_))
```

```python
def secant(f, guess, delta=1, tolerance = 2**-32):
    x0 = guess
    x1 = x0 + delta
    n = 0
    while abs(f(x1))>tolerance:
        x2 = x1 - (x1-x0)/(f(x1)-f(x0))*f(x1)
        x0 = x1
        x1 = x2
        n += 1
    return(x1, n)
```

```python
secant(f, 1000000, 100)
```
