Create a function and then use central difference to calculate the derivative of the function in the range of -2<x<2.

$$f(x) = 1+ \frac{1}{2}\tanh(2 x)$$

and the analytic derivative of that is:

$$
\DeclareMathOperator{\sech}{sech}
f'(x) =  \sech^2(2 x) = \frac{1}{\cosh^2(2 x)}
$$

Python doesn't have `sech` built in but it does have `cosh` so we'll just use that form.


```python
import numpy as np
import matplotlib.pyplot as plt
```

## First let's plot the function!

```python
def f(x):
    return(1+1/2*np.tanh(2*x))

x = np.linspace(-2,2)

# plot
fig, ax = plt.subplots()
ax.plot(x, f(x))
```

## Now we'll use our midpoint derivative formula!

```python
def mid(function,x_value, stepSize):
    return (function(x_value+stepSize)-function(x_value-stepSize))/(2*stepSize)

derivative = []
for i in x:
    f_prime = mid(f, i, 0.001)
    derivative.append(f_prime)

fig1, ax1 = plt.subplots()
ax1.plot(x, f(x))
ax1.plot(x, derivative, 'o', mfc='none')
ax1.plot(x, 1/np.cosh(2*x)**2) # the exact answer
ax1.legend([r'$f\,(x)$', r'$\frac{df}{dx}$ - computation', r'$\frac{df}{dx}$ - exact'])
```
